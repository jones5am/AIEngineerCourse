```python
# Imports
import os
from openai import OpenAI
from dotenv import load_dotenv
load_dotenv()
```

```python
# Establish connection to OpenAI
api_key = os.getenv("OPENAI_API_KEY")
client = OpenAI(api_key=api_key)
```

```python
# Create a response object
responses = client.responses
```

```python
# Example request demonstrating a continuous, multi-turn conversation:
input_data = [
    # 1. High-level rules for the model
    {"role": "developer", "content": "You are a helpful assistant that answers clearly and concisely."},
    
    # 2. First turn of the conversation (History)
    {"role": "user", "content": "Write a short greeting for someone new to the Fifa world cup"},
    
    # 3. This is what the assistant previously answered (Context)
    {"role": "assistant", "content": "Welcome to the FIFA World Cup! Get ready for an exciting journey filled with thrilling matches, passionate fans, and unforgettable moments. Enjoy the games and cheer for your favorite teams!"},
    
    # 4. The new active request building on top of the history
    {"role": "user", "content": "Can you translate that last sentence to Spanish?"}
]

response = responses.create(model="gpt-5.4-mini", input=input_data)
print(response.output[0].content[0].text)
```
