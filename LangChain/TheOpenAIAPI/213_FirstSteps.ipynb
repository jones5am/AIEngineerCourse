```python
from dotenv import load_dotenv
load_dotenv()
```

```python
import os
from openai import OpenAI
```

```python
api_key = os.getenv("OPENAI_API_KEY")
client = OpenAI(api_key=api_key)
```

```python
responses = client.responses
```

```python
# Example request structure for later use:
response = responses.create(model="gpt-5.4-nano", input="Which football player has scored the most goals in the world cup historically?")
print(response.output[0].content[0].text)
```
