```python
from dotenv import load_dotenv
load_dotenv()
```

```python
import os
from openai import OpenAI
```

```python
api_key = os.getenv("OPENAI_API_KEY")
client = OpenAI(api_key=api_key)
responses = client.responses
```

```python
response = responses.create(
    model="gpt-5.4-nano",
    input="Write a short paragraph about the most dramatic fictional narrative of the 2042 FIFA World Cup final.",
    temperature=0.01
)
print(response.output_text)
```

```python
response = responses.create(
    model="gpt-5.4-nano",
    input="Write a short paragraph about the most dramatic fictional narrative of the 2042 FIFA World Cup final.",
    temperature=1.9
)
print(response.output_text)
```

```python
response = responses.create(
    model="gpt-5.4-nano",
    input="Write a short paragraph about the most dramatic fictional narrative of the 2042 FIFA World Cup final.",
    temperature=0.01,
    max_output_tokens=60
)
print(response.output_text)
```

```python
import time
```

```python
stream = responses.create(
    model="gpt-5.4-nano",
    input=[
        {
            "role": "user",
            "content": "List three memorable FIFA World Cup moments from the past decade.",
        },
    ],
    stream=True,
)

for event in stream:
    if "output_text.delta" in event.type:
        print(event.delta, end="", flush=True)
        time.sleep(0.1)  # Simulate processing time for demonstration purposes
```

```python
response = responses.create(
    model="gpt-5.4-nano",
    input="In the FIFA World Cup group stage, Team A scored 5 goals and conceded 2, while Team B scored 4 goals and conceded 1. Which team has the better goal difference, and what is the difference? Show your reasoning steps.",
    reasoning={"effort": "high"}
    #max_output_tokens=360
)
print(response.output_text)
```

```python
chat_completions = client.chat.completions
```
