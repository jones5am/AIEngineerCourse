```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from datetime import date

from langchain_litellm import ChatLiteLLM
from langchain_core.messages import HumanMessage
from pydantic import BaseModel, Field
```

```python
llm = ChatLiteLLM(model="gpt-5.4-mini", temperature = 0, model_kwargs = {'seed':365})
```

```python
human_message = HumanMessage(content="Give the start dates of the past 3 Fifa World Cups")
```

```python
completion = llm.invoke([human_message])
print(completion.content)
```

```python
class WorldCupStartDate(BaseModel):
    """One FIFA World Cup and its opening match date."""

    start_date: date = Field(description="Opening match date (date only, no time)")


class PastThreeWorldCupStartDates(BaseModel):
    """Start dates for the past three FIFA World Cups."""

    world_cups: list[WorldCupStartDate] = Field(
        min_length=3,
        max_length=3,
        description="Past 3 FIFA World Cups, most recent first",
    )
```

```python
llm_with_structured_output = llm.with_structured_output(PastThreeWorldCupStartDates)
```

```python
completion = llm_with_structured_output.invoke([human_message])
```
