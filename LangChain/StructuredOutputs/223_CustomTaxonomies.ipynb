```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from enum import StrEnum

from langchain_litellm import ChatLiteLLM
from langchain_core.messages import HumanMessage
from pydantic import BaseModel, Field
```

```python
llm = ChatLiteLLM(model="gpt-5.4-mini", temperature = 0, model_kwargs = {'seed':365})
```

```python
human_message = HumanMessage(content="What European countries have won the Fifa World Cup?")
```

```python
class EuropeanCountryISOCode(StrEnum):
    """ISO 3166-1 alpha-2 codes for European countries."""

    AD = "AD"  # Andorra
    AL = "AL"  # Albania
    AT = "AT"  # Austria
    BA = "BA"  # Bosnia and Herzegovina
    BE = "BE"  # Belgium
    BG = "BG"  # Bulgaria
    BY = "BY"  # Belarus
    CH = "CH"  # Switzerland
    CY = "CY"  # Cyprus
    CZ = "CZ"  # Czechia
    DE = "DE"  # Germany
    DK = "DK"  # Denmark
    EE = "EE"  # Estonia
    ES = "ES"  # Spain
    FI = "FI"  # Finland
    FR = "FR"  # France
    GB = "GB"  # United Kingdom
    GR = "GR"  # Greece
    HR = "HR"  # Croatia
    HU = "HU"  # Hungary
    IE = "IE"  # Ireland
    IS = "IS"  # Iceland
    IT = "IT"  # Italy
    LI = "LI"  # Liechtenstein
    LT = "LT"  # Lithuania
    LU = "LU"  # Luxembourg
    LV = "LV"  # Latvia
    MC = "MC"  # Monaco
    MD = "MD"  # Moldova
    ME = "ME"  # Montenegro
    MK = "MK"  # North Macedonia
    MT = "MT"  # Malta
    NL = "NL"  # Netherlands
    NO = "NO"  # Norway
    PL = "PL"  # Poland
    PT = "PT"  # Portugal
    RO = "RO"  # Romania
    RS = "RS"  # Serbia
    RU = "RU"  # Russia
    SE = "SE"  # Sweden
    SI = "SI"  # Slovenia
    SK = "SK"  # Slovakia
    SM = "SM"  # San Marino
    TR = "TR"  # Turkey
    UA = "UA"  # Ukraine
    VA = "VA"  # Vatican City
    XK = "XK"  # Kosovo


class EuropeanWorldCupWinners(BaseModel):
    """European ISO country codes."""

    countries: list[EuropeanCountryISOCode] = Field(
        description="ISO 3166-1 alpha-2 codes of European countries",
    )
```

```python
llm_with_structured_output = llm.with_structured_output(EuropeanWorldCupWinners)
```

```python
completion = llm_with_structured_output.invoke([human_message])
```

```python
print(completion.countries)
```
