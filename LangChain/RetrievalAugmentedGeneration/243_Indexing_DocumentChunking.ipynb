```python
import docling
from pathlib import Path

from docling.backend.pypdfium2_backend import PyPdfiumDocumentBackend
from docling.datamodel.base_models import InputFormat
from docling.datamodel.pipeline_options import PdfPipelineOptions
from docling.document_converter import DocumentConverter, PdfFormatOption
from langchain_docling.loader import DoclingLoader, ExportType

from langchain_text_splitters.character import CharacterTextSplitter
from langchain_text_splitters.markdown import MarkdownHeaderTextSplitter
```

```python
# Get the file path to the PDF file
FILE_PATH = Path.cwd() / "RegulationsfortheFIFAWorldCup26_noimages.pdf"

# Set your pipeline options for docling
pipeline_options = PdfPipelineOptions(
    allow_external_plugins=True,
    do_ocr=False,
    do_table_structure=True,
    force_backend_text=True,
    generate_parsed_pages=False,
)

converter = DocumentConverter(
    format_options={
        InputFormat.PDF: PdfFormatOption(
            pipeline_options=pipeline_options,
            backend=PyPdfiumDocumentBackend,
        )
    }
)

loader = DoclingLoader(
    file_path=str(FILE_PATH),
    converter=converter,
    export_type=ExportType.MARKDOWN,
)

docs = loader.load()
markdown = docs[0].page_content

print("documents loaded:", len(docs))
print("markdown length:", len(markdown))
```

Chunk by Character Split

```python
char_splitter = CharacterTextSplitter(separator = ".", 
                                      chunk_size = 1000, 
                                      chunk_overlap = 50)
```

```python
pages_char_split = char_splitter.split_documents(docs)
```

```python
len(pages_char_split)
```

```python
print(pages_char_split[0].page_content)
```

Chunk by Markdown Headings

```python
md_splitter = MarkdownHeaderTextSplitter(headers_to_split_on = [("#", "Header 1"), 
                                                                ("##", "Header 2")])
```

```python
pages_md_split = md_splitter.split_text(docs[0].page_content)
```
