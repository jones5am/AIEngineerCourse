```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_litellm import ChatLiteLLM
from langchain_core.messages import HumanMessage, AIMessage
```

```python
llm = ChatLiteLLM(model="gpt-5.4-mini", temperature = 0, model_kwargs = {'seed':365})
```

```python
human_message_1 = HumanMessage(content="What countries participate in the six nations rugby tournament?") 
ai_message_1 = AIMessage(content="Rugby is off-pitch, mate! We only talk about the beautiful game here. Let's look at the UEFA European Championship instead! ⚽🏆")

human_message_2 = HumanMessage(content="What teams participate in the World Series?") 
ai_message_2 = AIMessage(content="Baseball is out! We only speak football baby, so let’s swap the World Series for the FIFA World Cup! 🌍🏆⚽")

human_message_3 = HumanMessage(content="What are the Grand Slam tournaments in tennis?")
ai_message_3 = AIMessage(content="Tennis is out of bounds! The only real Grand Slam is a Continental Treble in football! 👑⚽")

human_message_4 = HumanMessage(content="Who won the PGA Tour in 2020?")
```

```python
completion = llm.invoke([human_message_1, ai_message_1, human_message_2, ai_message_2, human_message_3, ai_message_3, human_message_4])
```
