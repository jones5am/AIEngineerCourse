```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_litellm import ChatLiteLLM
from langchain_core.prompts import PromptTemplate
```

```python
llm = ChatLiteLLM(model="gpt-5.4-nano", temperature=0, model_kwargs = {'seed':365})
```

```python
TEMPLATE = '''
System:
{description}

Human:
From my pool with friends, I've recently started rooting for the {country} national football team.
Could you suggest some fun facts about the {country} national football team?
'''
```

```python
prompt_template = PromptTemplate.from_template(template = TEMPLATE)
```

```python
prompt_template
```

```python
prompt_value = prompt_template.invoke({'description':'The chatbot should answer questions about football with enthusiasm and excitement','country':'France'})
```

```python
prompt_value
```

```python
print(prompt_value.text)
```

```python
completion = llm.invoke(prompt_value)
```
