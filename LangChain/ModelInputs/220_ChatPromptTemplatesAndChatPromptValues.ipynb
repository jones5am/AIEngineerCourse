```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_litellm import ChatLiteLLM

from langchain_core.prompts.chat import (SystemMessagePromptTemplate,
                                         HumanMessagePromptTemplate,
                                         ChatPromptTemplate)
```

```python
llm = ChatLiteLLM(model="gpt-5.4-mini", temperature = 0, model_kwargs = {'seed':365})
```

```python
TEMPLATE_S = '{description}'
TEMPLATE_H = '''From my pool with friends, I've recently started rooting for the {country} national football team.
Could you suggest some fun facts about the {country} national football team?'''

message_template_s = SystemMessagePromptTemplate.from_template(template = TEMPLATE_S)
message_template_h = HumanMessagePromptTemplate.from_template(template = TEMPLATE_H)
```

```python
message_template_h
```

```python
chat_template = ChatPromptTemplate.from_messages([message_template_s, message_template_h])
```

```python
chat_template
```

```python
chat_value = chat_template.invoke({'description':'The chatbot should answer questions about football with enthusiasm and excitement','country':'Spain'})
```

```python
chat_value
```

```python
completion = llm.invoke(chat_value)
```
