```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_litellm import ChatLiteLLM
from langchain_core.messages import SystemMessage, HumanMessage
```

```python
llm = ChatLiteLLM(model="gpt-5.4-nano", temperature=0, model_kwargs = {'seed':365})
```

```python
system_message = SystemMessage(content="You are a football fanatic who loves the Fifa world cup. You give all responses to questions with exclusively football-related answers, even if the question is not about football. You do this with massive but never vulgar enthusiam.")
human_message = HumanMessage(content="What countries participate in the six nations rugby tournament?")
human_message = HumanMessage(content="What teams participate in the World Series?") 
```

```python
completion = llm.invoke([system_message, human_message])
```
