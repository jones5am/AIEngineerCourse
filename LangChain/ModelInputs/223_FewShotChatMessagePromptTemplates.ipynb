```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_litellm import ChatLiteLLM

from langchain_core.prompts import (ChatPromptTemplate,
                                    HumanMessagePromptTemplate, 
                                    AIMessagePromptTemplate, 
                                    FewShotChatMessagePromptTemplate)
```

```python
llm = ChatLiteLLM(model="gpt-5.4-mini", temperature = 0, model_kwargs = {'seed':365})
```

```python
TEMPLATE_H = '''From my betting pool with friends, I've recently started rooting for the {country} national team.
Could you suggest some fun facts about the {country} national team?'''
TEMPLATE_AI = '''{response}'''

message_template_h = HumanMessagePromptTemplate.from_template(template = TEMPLATE_H)
message_template_ai = AIMessagePromptTemplate.from_template(template = TEMPLATE_AI)
```

```python
example_template = ChatPromptTemplate.from_messages([message_template_h, 
                                                     message_template_ai])
```

```python
examples = [
    {'country':'Brazil',
     'response':'''Brazil is football royalty in the World Cup. From Pele's 1970 artistry to the 2002 golden generation, their samba style and trophy pedigree make them an automatic conversation about beautiful football.'''} ,
    {'country':'Germany',
     'response':'''Germany is built on classic World Cup football: efficient tactics, late goals, and four championship trophies. Their 2014 semifinal demolition of Brazil is a signature football moment that only a World Cup obsession would savor.'''} ,
    {'country':'Argentina',
     'response':'''Argentina is pure World Cup drama and football soul. Maradona's '86 magic and Messi's 2022 triumph are the kind of stories that keep every football fan talking about the tournament.'''} ,
    {'country':'Senegal',
     'response':'''Senegal represents the rising power of World Cup football from Africa. Their 2002 quarterfinal run and fearless attack prove that the tournament is about global football passion, not just the usual giants.'''} ,
    {'country':'Japan',
     'response':'''Japan brings sharp, well-drilled football to the World Cup. Their pace, disciplined counters, and tournament resilience are exactly the kind of football details that make the World Cup fascinating.'''} ,
    {'country':'England',
     'response':'''England is all about World Cup football tradition and history. With iconic moments like 1966 and recent knockout drama, their football story is one that always circles back to the tournament.'''}]
```

```python
few_shot_prompt = FewShotChatMessagePromptTemplate(examples = examples, 
                                                   example_prompt = example_template, 
                                                   input_variables = ['country'])
```

```python
chat_template = ChatPromptTemplate.from_messages([few_shot_prompt, 
                                                  message_template_h])
```

```python
chat_value = chat_template.invoke({'country':'Fiji'})
```

```python
chat_value
```

```python
for i in chat_value.messages:
    print(f'{i.type}: {i.content}\n')
```

```python
completion = llm.invoke(chat_value)
```

```python
print(completion.content)
```
