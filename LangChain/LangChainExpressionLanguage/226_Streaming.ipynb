```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_litellm import ChatLiteLLM
from langchain_core.prompts import ChatPromptTemplate
```

```python
chat_template = ChatPromptTemplate.from_messages([
    ('human',
     "I'm researching the {year} FIFA World Cup and the {team} national team. Could you suggest several memorable moments from that tournament?")])
```

```python
llm = ChatLiteLLM(
    model="gpt-5.4-mini",
    temperature=0,
    model_kwargs={'seed': 365},
    max_tokens=100,
    streaming=True,
)
```

```python
chain = chat_template | llm
chunk_count = 0
```

```python
for chunk in chain.stream({'year': '2018', 'team': 'France'}):
    print(chunk.content, end='', flush=True)
    chunk_count += 1
```

```python
print('Count of Chunks: ', chunk_count)
```
