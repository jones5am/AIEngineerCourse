```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_litellm import ChatLiteLLM
from langchain_core.prompts import ChatPromptTemplate
from pydantic import BaseModel, Field
```

```python
class MemorableItems(BaseModel):
    """Three memorable items related to the FIFA World Cup research topic."""

    items: list[str] = Field(
        min_length=3,
        max_length=3,
        description="Three memorable items",
    )
```

```python
MemorableItems.model_json_schema()
```

```python
chat_template = ChatPromptTemplate.from_messages([
    ('human',
     "I'm researching the {year} FIFA World Cup. Could you suggest three memorable {topic}?")])
```

```python
print(chat_template.messages[0].prompt.template)
```

```python
llm = ChatLiteLLM(model="gpt-5.4-mini", temperature=0, model_kwargs={'seed': 365})
```

```python
llm_with_structured_output = llm.with_structured_output(MemorableItems)
```

Without a chain

```python
chat_template_result = chat_template.invoke({'year': '2022', 'topic': 'goals'})
```

```python
chat_result = llm_with_structured_output.invoke(chat_template_result)
```

```python
chat_result.items
```

With a chain

```python
chain = chat_template | llm_with_structured_output
```

```python
chain_result = chain.invoke({'year': '2022', 'topic': 'goals'})
```

```python
chain_result.items
```
