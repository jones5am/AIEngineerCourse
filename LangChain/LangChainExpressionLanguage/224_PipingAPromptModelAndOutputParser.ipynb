```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_litellm import ChatLiteLLM
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import CommaSeparatedListOutputParser
```

```python
list_instructions = CommaSeparatedListOutputParser().get_format_instructions()
```

```python
list_instructions
```

```python
chat_template = ChatPromptTemplate.from_messages([
    ('human',
     "I'm researching the {year} FIFA World Cup. Could you suggest three memorable {topic}? \n" + list_instructions)])
```

```python
print(chat_template.messages[0].prompt.template)
```

```python
llm = ChatLiteLLM(
    model="gpt-5.4-mini",
    temperature=0,
    model_kwargs={'seed': 365},
    max_tokens=100,
)
```

```python
list_output_parser = CommaSeparatedListOutputParser()
```

```python
chat_template_result = chat_template.invoke({'year': '2022', 'topic': 'goals'})
```

```python
chat_result = llm.invoke(chat_template_result)
```

```python
list_output_parser.invoke(chat_result)
```

```python
chain = chat_template | llm | list_output_parser
```

```python
chain.invoke({'year': '2022', 'topic': 'goals'})
```
